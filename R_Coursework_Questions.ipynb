The `dplyr` dataset `storms` is part of the US National Oceanic and Atmospheric Administration database of Atlantic hurricanes. It contains multiple repeated measurements from hurricanes. For a given storm, there are measurements from every 6 hours across the life time of the storm.

Use a tidyverse approach with `dplyr` to answer the following:

```r
#install.packages("tidyverse")
#install.packages("dplyr")
library(tidyverse)
library(dplyr)

#calculating number of different storms
unique_storms <- tally(distinct(storms, name, .keep_all = FALSE))
print(paste0('The number of different storms is: ', unique_storms))

#calculating which year had the most storms
unique_storms <- distinct(storms, name, .keep_all = TRUE)
unique_storms %>%
  count(year) %>%
  arrange(desc(n)) %>%
  head(1)

#calculating which storm, in which year had the highest pressure
storms %>%
  select(name, year, pressure) %>%
  arrange(pressure) %>%
  head(1)
```

### Question 8

The `ggplot2` dataset `diamonds` contains the price and information from the sale of 54,000 different diamonds. You can use ?diamonds to find out more about the dataset if you want.

Using `ggplot2`:

- plot the relationship between the carat (i.e. the weight of the diamond) and the price.
- plot a histogram of prices of all the diamonds
- plot a boxplot of the difference in price between the different cuts

```r
#install.packages("tidyverse")
library(ggplot2)

#plot relationship between carat and price
ggplot(diamonds, (aes(carat, price))) +
  geom_point(alpha = 0.2, colour = "blue", fill = "black", 
             shape = 21, size = 1) +
  labs(title = "Scatterplot showing Carat vs Price",
       x = "Carat", 
       y = "Price ($)")


#histogram of prices of all diamonds
ggplot(diamonds, (aes(x=price))) +
  geom_histogram(bins = 15, fill = "cadetblue2", color = "grey") +
  labs(title = "Histogram of Prices",
       x = "Price ($)",
       y = "Number of Diamonds")

#boxplot of price variation between cuts
diamonds %>%
  ggplot(aes(x = cut, y = price, fill = cut)) +
  geom_boxplot()+ 
  labs(title = "Boxplot Showing Price Variation Between Cuts",
       x = "Cut Type",
       y = "Price ($)")
```

### Question 10

Sticking with the `diamonds` dataset for this question. A carat is a special unit of weight for gemstones where a single carat is 200 mg or 0.2 g. Write a function to convert the `diamonds$carat` vector into grams or milligrams. How heavy is the heaviest diamond?

```r
#install.packages("tidyverse")
#install.packages("dplyr")
library(tidyverse)
library(dplyr)

#converts carat weight into grams or mg
carat_weight_converter <- function(unit){
  if(tolower(unit) == "grams"){
    diamonds %>%
      mutate(carat = carat * 0.2) %>% #converts weight from carats into grams
      rename(Grams = carat) %>%
      head()
  }
  
  else if (tolower(unit) == "mg"){
    diamonds %>%
      mutate(carat = carat * 200) %>% #converts weight from carats into mg
      rename(Mg = carat) %>%
      head()
  }
  
  else{
    print("You must enter 'grams' or 'mg'")
  }
  
}

carat_weight_converter() #enter 'grams' or 'mg' to retrieve the respective weight

#finds the heaviest diamond in either grams, mg or carat
heaviest_diamond <- function(unit){
  if(tolower(unit) == "grams"){
    weight_grams <- diamonds$carat * 0.2 #converts weight from carats into grams
    heaviest_grams <- max(weight_grams) #finds the heaviest diamond
    return(print(paste("The heaviest diamond in grams is: ", heaviest_grams, 'g.', sep = '')))
  }
  
  else if (tolower(unit) == "mg"){
    weight_mg <- diamonds$carat * 200 #converts weight from carats into mg
    heaviest_mg <- max(weight_mg) #finds the heaviest diamond
    return(print(paste("The heaviest diamond in mg is: ", heaviest_mg, 'mg.', sep = '')))
  }
  
  else if (tolower(unit) == "carat"){
    heaviest_carat <- max(diamonds$carat) #finds the heaviest diamond
    return(print(paste("The heaviest diamond in carats is: ", heaviest_carat, ' carats.', sep = '')))
  }
  
  else{
    print("You must enter 'grams', 'mg' or 'carat'")
  }
  
}

heaviest_diamond() #enter 'grams', 'mg' or 'carat' to retrieve the respective weight
```
