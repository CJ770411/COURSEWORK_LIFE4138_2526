### Question 7
Analyze the life expectancy dataset (which can be loaded by running the cell below).

- Calculate the number of unique countries in the dataset.

- Calculate the country and year with the highest life expectancy.

- Calculate the year with the highest total expenditure.

- Provide your code using pandas.


````python
import seaborn as sns
life_expectancy = sns.load_dataset('healthexp')
````

````python
import pandas as pd
import seaborn as sns
life_expectancy = sns.load_dataset('healthexp')

#1 calculating number of unique countries
unique_countries_count = life_expectancy['Country'].nunique() #counts the amount of unique countries
print("The amount of unique countries is:", unique_countries_count, "\n")

#2 calculating country and year with highest life expectancy
country = life_expectancy[['Country', 'Life_Expectancy']] #selects only the Country and Life Expectancy columns from the dataset
country_df = pd.DataFrame(country) #converts the dataset to a pandas dataframe
sorted_country_df = country_df.sort_values(by='Life_Expectancy', ascending=False) #arranges the dataframe into descending order based on Life Expectancy

print("The country with the highest life expectancy is:\n", sorted_country_df.head(1), "\n")

year = life_expectancy[['Year', 'Life_Expectancy']] #selects only the Year and Life Expectancy columns from the dataset
year_df = pd.DataFrame(year) #converts the dataset to a pandas dataframe
sorted_year_df = year_df.sort_values(by='Life_Expectancy', ascending=False) #arranges the dataframe into descending order based on Life Expectancy

print("The year with the highest life expectancy is:\n", sorted_year_df.head(1), "\n")

#for both Country and Year with highest life expectancy presented together
country_year = life_expectancy[['Country', 'Year', 'Life_Expectancy']] #selects only the Country, Year and Life Expectancy columns from the dataset
country_year_df = pd.DataFrame(country_year) #converts the dataset to a pandas dataframe
sorted_country_year_df = country_year_df.sort_values(by='Life_Expectancy', ascending=False) #arranges the dataframe into descending order based on Life Expectancy

print("The country and year with the highest life expectancy presented together is:\n", sorted_country_year_df.head(1), "\n")

#3 calculating the year with the highest total expenditure
expenditure = life_expectancy.groupby('Year')['Spending_USD'].sum() #selects only the Year and Spending_USD columns from the dataset and sums the total Spending_USD for each year
expenditure_df = pd.DataFrame(expenditure) #converts the dataset to a pandas dataframe
sorted_expenditure_df = expenditure_df.sort_values(by='Spending_USD', ascending=False) #arranges the dataframe into descending order based on Spending_USD
print("The year with the highest total expenditure was:\n", sorted_expenditure_df.head(1))
````

````python
import seaborn as sns
import matplotlib.pyplot as plt

diamonds = sns.load_dataset('diamonds')

#1 Carat vs Price in a scatter plot
print("Here is my scatterplot showing Carat vs Price:\n")
diamonds.plot.scatter(x = 'carat', y = 'price', alpha = 0.3, color = 'orange')
plt.title('Scatterplot showing Carat vs Price')
plt.xlabel('Carat')
plt.ylabel('Price ($)')
plt.show()

#2 Histogram of prices
diamonds = sns.load_dataset('diamonds')
print("Here is my histogram of Prices:\n")

plt.hist(diamonds['price'], bins=15, fill = False)
plt.title('Histogram of prices')
plt.xlabel('Price ($)')
plt.ylabel('Frequency')
plt.show()

#3 Boxplot of cut vs price
diamonds = sns.load_dataset('diamonds')
print("Here is my boxplot of Cut vs Price:\n")

sns.boxplot(x = 'cut', y = 'price', data = diamonds, color = 'skyblue')
plt.title('Boxplot showing Cut vs Price')
plt.xlabel('Cut')
plt.ylabel('Price ($)')
plt.show()
````

### Question 10
Write a function that will convert carat to grams and find the heaviest diamond.

1 carat = 0.2 grams

````python
import seaborn as sns
import pandas as pd

diamonds = sns.load_dataset('diamonds')

def diamonds_weight_converter():
    diamonds_df = pd.DataFrame(diamonds) #convert diamonds to a pandas dataframe
    diamonds_grams = diamonds_df.rename(columns={'carat': 'Grams(g)'}, inplace=False) #renames the carat column to grams
    diamonds_grams.loc[:, 'Grams(g)'] = diamonds_grams['Grams(g)'] * 0.2 #converts carats to grams
    print(diamonds_grams.head())
    print("\nThe heaviest diamonds weighs: ", max(diamonds_grams['Grams(g)']), "g", sep = '')

diamonds_weight_converter()
````
